Projeto de Mercado Financeiro

Etapas:
1) Importar as bibliotecas yfinance, pandas, datetime e plotly.
2) Importar os indices do yfinance IBOVESPA, DOLAR
4) Colocar no gráfico todos indices e fazer a correlação.
5) Analisar periodo da pandemia.

```python
import yfinance as yf
import pandas as pd
from datetime import datetime
import plotly.graph_objects as go
from plotly.subplots import make_subplots

data_inicio = '2000-01-01'
data_fim = '2026-12-31'

indices = ['^BVSP', 'BRL=X']

def pegar_cotacoes(lista_tickers, data_inicio, data_fim):
    df = yf.download(
        lista_tickers,
        start=data_inicio,
        end=data_fim,
        auto_adjust=False
    )
    df = df['Adj Close']
    df = df.ffill()
    return df

lista_ticker = indices
df_cotacoes = pegar_cotacoes(lista_ticker, data_inicio, data_fim)
display(df_cotacoes)
```

```python
df_grafico = df_cotacoes[['^BVSP', 'BRL=X']].dropna()

grafico = make_subplots(specs=[[{"secondary_y": True}]])

grafico.add_trace(go.Scatter(
    x=df_grafico.index,
    y=df_grafico['^BVSP'],
    name='IBOVESPA'
), secondary_y=False)

grafico.add_trace(go.Scatter(
    x=df_grafico.index,
    y=df_grafico['BRL=X'],
    name='Dólar'
), secondary_y=True)

grafico.update_layout(
    template='plotly_dark',
    title='IBOVESPA x Dólar - Preços reais',
    hovermode='x unified'
)

grafico.update_yaxes(
    title_text='IBOVESPA',
    secondary_y=False
)

grafico.update_yaxes(
    title_text='Dólar BRL',
    secondary_y=True
)

grafico.show()
```

```python
# Periodo da pandemia
data_pandemia = '2020-01-01'
fim_pandemia = '2020-12-31'

df_cotacoes = pegar_cotacoes(lista_ticker, data_pandemia, fim_pandemia)
display(df_cotacoes)
```

```python
df_pandemia = df_cotacoes[['^BVSP', 'BRL=X']].dropna()
grafico = make_subplots(specs=[[{"secondary_y": True}]])

grafico.add_trace(go.Scatter(
    x=df_pandemia.index,
    y=df_pandemia['^BVSP'],
    name='IBOVESPA'
), secondary_y=False)

grafico.add_trace(go.Scatter(
    x=df_pandemia.index,
    y=df_pandemia['BRL=X'],
    name='Dólar'
), secondary_y=True)

grafico.update_layout(
    template='plotly_dark',
    title='IBOVESPA x Dólar - Durante a pandemia - 2020',
    hovermode='x unified'
)

grafico.update_yaxes(
    title_text='IBOVESPA',
    secondary_y=False
)

grafico.update_yaxes(
    title_text='Dólar BRL',
    secondary_y=True
)

grafico.show()
```

```python
df_pandemia = df_cotacoes[['^BVSP', 'BRL=X']].dropna()

ibov_inicio = df_pandemia['^BVSP'].iloc[0]
ibov_minimo = df_pandemia['^BVSP'].min()
data_ibov_minimo = df_pandemia['^BVSP'].idxmin()

dolar_inicio = df_pandemia['BRL=X'].iloc[0]
dolar_maximo = df_pandemia['BRL=X'].max()
data_dolar_maximo = df_pandemia['BRL=X'].idxmax()

queda_ibov = (ibov_minimo / ibov_inicio - 1) * 100
alta_dolar = (dolar_maximo / dolar_inicio - 1) * 100

print(f"Ibovespa caiu {queda_ibov:.2f}% até {data_ibov_minimo.date()}")
print(f"Dólar subiu {alta_dolar:.2f}% até {data_dolar_maximo.date()}")
```
